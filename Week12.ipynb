# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

In the graph above you can see the 2 plots side by side. What's good about this graph is that it separates the two plots yet you can still compare them. However, it isn't as "clean" as the next graph. It also doesn't use color to differentiate like the other graph does. But if there were more weeks then this graph would be easier to comprehend. 

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

What's great about this graph is that you can compare the two plots directly since they're part of the same graph. The difference in colors makes it easier to tell which week is being represented. You could also need a legend here to fully understand it otherwise the viewer wouldn't be able to see which graph is representing which week. Personally, I think this is the cleaner and more "professional-looking" visulization. It's simple and easy to comprehend. However, if more weeks need to represented in the future then this would get too crowded.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

I will be reproducing the retail price over time graph. 

```python

from scipy.interpolate import make_interp_spline

years = np.array([2000, 2002, 2004, 2006, 2008, 2010, 2015, 2020, 2024])

a_vals = np.array([4.2, 4.5, 4.7, 5.0, 5.1, 4.3, 3.2, 2.6, 2.4])
b_vals = np.array([3.9, 4.2, 4.4, 4.7, 4.8, 4.0, 3.0, 2.45, 2.2])

# smooth with cubic spline
x_smooth = np.linspace(years.min(), years.max(), 400)
a_smooth = make_interp_spline(years, a_vals)(x_smooth)
b_smooth = make_interp_spline(years, b_vals)(x_smooth)

fig, ax = plt.subplots(figsize=(10, 4))

navy = "#1a3f6f"
ax.plot(x_smooth, a_smooth, color=navy, linewidth=2.2)
ax.plot(x_smooth, b_smooth, color=navy, linewidth=2.2)

# labels on left side
ax.text(years[0] - 0.8, a_vals[0],  "A", va="center", ha="right",
        color="#666666", fontsize=11)
ax.text(years[0] - 0.8, b_vals[0],  "B", va="center", ha="right",
        color="#666666", fontsize=11)

ax.set_xlabel("Year", fontsize=10)
ax.set_ylabel("Average price ($)", fontsize=10)


ax.tick_params(bottom=False)
ax.set_xticks([])

plt.tight_layout()
plt.show()
```
